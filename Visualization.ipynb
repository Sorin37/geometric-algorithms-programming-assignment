```python
H = [(1, 1), (5, 5), (2, 4), (9, -3), (12, 4)]
P = [(3, 2), (6, -1), (3, 3)]
d = 2
```

```python
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.patches as patches
from matplotlib.collections import PatchCollection

fig, ax = plt.subplots()
# houses
x = []
y = []

for h_x, h_y in H:
    x.append(h_x)
    y.append(h_y)
    
ax.scatter(x, y, c='brown', label='House')
# ax.scatter(x, y, c='#970125', label='House region', alpha=0.5, edgecolors='red', marker='s', s=d * 1000)
squares = [patches.Rectangle((x - d, y - d), width=d*2, height=d*2) for x, y in H]
collection = PatchCollection(
    squares, facecolor="#970125", edgecolor="red", alpha=0.5
)
ax.add_collection(collection)

# particles

x = []
y = []

for p_x, p_y in P:
    x.append(p_x)
    y.append(p_y)

ax.scatter(x, y, c='green', label='Particle')


ax.set_aspect("equal")
ax.autoscale_view()
ax.legend()
ax.grid(True)

leg = plt.legend()
for handle in leg.legend_handles:
    handle.set_sizes([36])

plt.show()
```
